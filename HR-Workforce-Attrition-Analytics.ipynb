```python
# EDA = Exploratory Data Analysis
```

```python
# EDA 1 - Attrition distribution
```

```python
sns.countplot(data=df,x="attrition")
plt.title("Employee Attrition Distribution")
plt.show()
```

```python
# EDA 2 - Attrition by department
```

```python
sns.countplot(data=df,x="department",hue="attrition")
plt.title("Attrition by Department")
plt.xticks(rotation=20)
plt.show()
```

```python
# EDA 3 - Attrition by overtime
```

```python
sns.countplot(data=df,x="over_time",hue="attrition")
plt.title("Attrition by overtime")
plt.show()
```

```python
# EDA 4 -  Attrition by age group
```

```python
sns.countplot(data=df,x="age_group", hue="attrition")
plt.title("Attrition  by AGe group")
plt.show()
```

```python
#  EDA 5 Job role analysis
```

```python
role_attrition = pd.crosstab(
    df["job_role"],
    df["attrition"]
)
role_attrition.plot(kind="bar", figsize=(12,6))
plt.title("Attrition by Job Role")
plt.xticks(rotation=45)
plt.show()
```

```python
# EDA 6 - Monthly Income
```

```python
sns.boxplot(data=df, x="attrition",y="monthly_income")
plt.title("Monthly Income vs Attrition")
plt.show()
```

```python
# EDA 7 - JOb satisfaction
```

```python
sns.countplot(
    data=df,
    x="job_satisfaction",
    hue="attrition"
)
plt.title("job Satisfaction vs Attrition")
plt.show()
```
